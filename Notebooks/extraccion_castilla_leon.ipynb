```python
# ============================================================
# 1. Cargar parquet de muestra Segovia/Madrid/restantes
# ============================================================

from pathlib import Path
import pandas as pd

ruta_silver = Path("../data/Silver")

ruta_parquet = ruta_silver / "muestra_20_segovia_18_madrid_2_restantes.parquet"

if not ruta_parquet.exists():
    raise FileNotFoundError(f"No existe el archivo: {ruta_parquet}")

df_muestra = pd.read_parquet(ruta_parquet)

print("Ruta cargada:", ruta_parquet)
print("Shape:", df_muestra.shape)

print("\nColumnas:")
print(df_muestra.columns.tolist())

display(df_muestra.head())
```

```python
# ============================================================
# 2. Revisar dominios de la muestra
# ============================================================

display(
    df_muestra["dominio_url"]
    .value_counts(dropna=False)
    .reset_index()
    .rename(columns={"index": "dominio_url", "dominio_url": "n"})
)
```

```python
df_muestra.columns
```

```python
# ============================================================
# 3. Filtrar Castilla y León
# ============================================================

df_castilla_leon = df_muestra[
    df_muestra["dominio_url"]
    .astype(str)
    .str.contains("contratacion.jcyl.es", case=False, na=False)
].copy()

print("Registros Castilla y León:", df_castilla_leon.shape)

display(
    df_castilla_leon[
        [
            "licitacion_id",
            "dominio_url",
            "detail_url",
            "url",
            "detail_url",
        ]
    ]
)
```

```python
# ============================================================
# 3. Revisar detail_url de Castilla y León
# ============================================================

df_castilla_leon = df_muestra[
    df_muestra["dominio_url"]
    .astype(str)
    .str.contains("contratacion.jcyl.es", case=False, na=False)
].copy()

print("Registros Castilla y León:", df_castilla_leon.shape)

display(
    df_castilla_leon[
        [
            "licitacion_id",
            "dominio_url",
            "detail_url"
        ]
    ]
)
```

```python
# ============================================================
# 4. Diagnóstico de detail_url Castilla y León
# ============================================================

import requests
from bs4 import BeautifulSoup
import pandas as pd
import re

headers = {
    "User-Agent": (
        "Mozilla/5.0 (Windows NT 10.0; Win64; x64) "
        "AppleWebKit/537.36 (KHTML, like Gecko) "
        "Chrome/120.0 Safari/537.36"
    ),
    "Accept-Language": "es-ES,es;q=0.9,en;q=0.8",
}

fila_castilla = df_castilla_leon.iloc[0].copy()

licitacion_id_castilla = fila_castilla["licitacion_id"]
detail_url_castilla = fila_castilla["detail_url"]

print("Licitación:", licitacion_id_castilla)
print("detail_url:", detail_url_castilla)

response_castilla = requests.get(
    detail_url_castilla,
    headers=headers,
    timeout=30,
    allow_redirects=True
)

print("Status:", response_castilla.status_code)
print("URL response:", response_castilla.url)
print("Content-Type:", response_castilla.headers.get("content-type"))
print("Tamaño respuesta:", len(response_castilla.content))
```

```python
# ============================================================
# 5. Detectar tipo de contenido de detail_url
# ============================================================

content_type = str(response_castilla.headers.get("content-type")).lower()

if (
    response_castilla.content.startswith(b"%PDF")
    or "application/pdf" in content_type
    or response_castilla.url.lower().endswith(".pdf")
):
    tipo_detectado_castilla = "pdf_directo"
elif "text/html" in content_type or "<html" in response_castilla.text.lower():
    tipo_detectado_castilla = "html"
elif "xml" in content_type or response_castilla.content.startswith(b"<?xml"):
    tipo_detectado_castilla = "xml"
else:
    tipo_detectado_castilla = "otro"

print("Tipo detectado:", tipo_detectado_castilla)
```

```python
# ============================================================
# 6. Si es HTML, revisar líneas extraídas
# ============================================================

if tipo_detectado_castilla == "html":
    soup_castilla = BeautifulSoup(response_castilla.text, "html.parser")

    texto_castilla = soup_castilla.get_text("\n", strip=True)

    lineas_castilla = [
        re.sub(r"\s+", " ", linea).strip()
        for linea in texto_castilla.split("\n")
        if re.sub(r"\s+", " ", linea).strip() != ""
    ]

    df_lineas_castilla = pd.DataFrame({
        "orden": range(len(lineas_castilla)),
        "linea": lineas_castilla
    })

    pd.set_option("display.max_rows", 200)
    pd.set_option("display.max_colwidth", 300)

    display(df_lineas_castilla.head(200))
else:
    print("No es HTML. No se extraen líneas.")
```
