```python
import pdfplumber

def extraer_pdf(ruta_pdf):

    texto = ""

    try:

        with pdfplumber.open(ruta_pdf) as pdf:

            for pagina in pdf.pages:

                texto += (
                    pagina.extract_text() or ""
                )

    except Exception as e:

        print(e)

    return texto
```

```python
from pathlib import Path

carpeta_base = Path("../data/Silver/Licitaciones")

registros = []

for pdf in carpeta_base.rglob("*.pdf"):

    texto = extraer_pdf(pdf)

    registros.append({
        "archivo": pdf.name,
        "ruta": str(pdf),
        "texto": texto
    })

pdfs = pd.DataFrame(registros)

pdfs.head()
```

```python
cols_emb = [c for c in dataset_final.columns if c.startswith("emb_")]

historicas = dataset_final[cols_emb].values
```

```python
from sklearn.metrics.pairwise import cosine_similarity

sim = cosine_similarity(
    embeddings_parquet,
    historicas
)
```

```python
print(sim.shape)
```

```python
import numpy as np

idx_max = np.argmax(sim, axis=1)

df_parquet["score_match"] = sim.max(axis=1)

df_parquet["historica_match"] = [
    dataset_final.iloc[i]["expediente"]
    for i in idx_max
]
```

```python
df_parquet["score_pct"] = (
    df_parquet["score_match"] * 100
).round(1)
```

```python
top3 = []

for fila in sim:

    idx = np.argsort(fila)[-3:][::-1]

    top3.append([
        {
            "expediente": dataset_final.iloc[i]["expediente"],
            "score": round(float(fila[i]*100),1)
        }
        for i in idx
    ])

df_parquet["top3_matches"] = top3
```

```python
df_parquet[
    [
        "titulo_original",
        "score_pct",
        "historica_match"
    ]
].sort_values(
    "score_pct",
    ascending=False
)
```
