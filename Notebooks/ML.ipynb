```python
import pdfplumber

def extraer_pdf(ruta_pdf):

    texto = ""

    try:

        with pdfplumber.open(ruta_pdf) as pdf:

            for pagina in pdf.pages:

                texto += (
                    pagina.extract_text() or ""
                )

    except Exception as e:

        print(e)

    return texto
```

```python
from pathlib import Path

carpeta_base = Path("../data/Silver/Licitaciones")

registros = []

for pdf in carpeta_base.rglob("*.pdf"):

    texto = extraer_pdf(pdf)

    registros.append({
        "archivo": pdf.name,
        "ruta": str(pdf),
        "texto": texto
    })

pdfs = pd.DataFrame(registros)

pdfs.head()
```

```python
def buscar_importes(texto):

    patron = r"\d{1,3}(?:\.\d{3})*(?:,\d+)?\s*€"

    return re.findall(
        patron,
        str(texto)
    )
```

```python
import re

def cpv_principal(texto):

    cpvs = re.findall(
        r"\d{8}-\d",
        str(texto)
    )

    if len(cpvs) > 0:
        return cpvs[0]

    return None
```
